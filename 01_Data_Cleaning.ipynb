###Lahore Air Quality & Smog Forecast

####Notebook 1: Data Cleaning
Purpose: Prepare the historical Lahore air quality dataset for machine learning by

*   loading the data,
*   checking data quality,
*   handling missing values,
*   formatting dates,

and preparing it for feature engineering.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("lahore_air_quality.csv")
```

```python
print(df.shape)
print(df.columns)
df.head()
```

```python
df = pd.read_csv(
    "lahore_air_quality.csv",
    engine="python"
)
```

```python
print("Number of stations:", df["name"].nunique())

stations = sorted(df["name"].unique())
print(stations)
```

```python
df["datetime_utc"] = pd.to_datetime(df["datetime_utc"])
df["date"] = df["datetime_utc"].dt.date
```

```python
daily_df = df.groupby(["name", "date"]).agg({
    "pm25": "mean",
    "temperature": "mean",
    "relativehumidity": "mean",
    "wind_speed": "mean",
    "wind_direction": "mean",
    "lat": "first",
    "lon": "first",
    "location_id": "first",
    "is_smog_season": "first"
}).reset_index()
```

```python
daily_df.isnull().sum()
```

```python
daily_df = daily_df.drop(columns=["pm10", "pm10_roll24h"], errors="ignore")
```

```python
print(daily_df.shape)
print(daily_df["name"].nunique())
daily_df.head()
```

```python
daily_df.to_csv("lahore_air_quality_cleaned.csv", index=False)
```
