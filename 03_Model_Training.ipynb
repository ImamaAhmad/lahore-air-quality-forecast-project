### Lahore Air Quality & Smog Forecast

### Notebook 3: Model Training

Purpose:

Train machine learning models to predict future PM2.5 concentrations in Lahore.

In this notebook, I will:
- Load the feature-engineered dataset.
- Select input features and prediction targets.
- Split the data into training and testing sets.
- Train Random Forest regression models.
- Save the trained models for future use.

The models developed here will later be used to generate AQI forecasts for the website.

```python
import pandas as pd
import joblib

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("lahore_air_quality_features.csv")
```

```python
print(df.shape)
df.head()
print(df.columns)
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df = df.sort_values(["date", "name"]).reset_index(drop=True)
```

```python
features = [
    "pm25",
    "temperature",
    "relativehumidity",
    "wind_speed",
    "wind_direction"
]
```

```python
X = df[features]
```

```python
y_day1 = df["pm25_day1"]
y_day2 = df["pm25_day2"]
y_day3 = df["pm25_day3"]
```

```python
split_date = df["date"].quantile(0.8)

train = df[df["date"] <= split_date]
test = df[df["date"] > split_date]

print("Training rows:", train.shape[0])
print("Testing rows:", test.shape[0])
```

```python
X_train = train[features]
X_test = test[features]

y1_train = train["pm25_day1"]
y1_test = test["pm25_day1"]

y2_train = train["pm25_day2"]
y2_test = test["pm25_day2"]

y3_train = train["pm25_day3"]
y3_test = test["pm25_day3"]
```

```python
model_day1 = RandomForestRegressor(
    n_estimators=50,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

model_day1.fit(X_train, y1_train)
```

```python
model_day2 = RandomForestRegressor(
    n_estimators=50,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

model_day2.fit(X_train, y2_train)
```

```python
model_day3 = RandomForestRegressor(
    n_estimators=50,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

model_day3.fit(X_train, y3_train)
```

```python
prediction_day1 = model_day1.predict(X_test)
prediction_day2 = model_day2.predict(X_test)
prediction_day3 = model_day3.predict(X_test)
```

```python
joblib.dump(model_day1, "pm25_day1_model.pkl")
joblib.dump(model_day2, "pm25_day2_model.pkl")
joblib.dump(model_day3, "pm25_day3_model.pkl")
```
