### Lahore Air Quality & Smog Forecast
#### Notebook 2: Feature Engineering

Purpose: Prepare the cleaned daily dataset for machine learning by:

*   creating prediction targets,
*   generating lag features from historical PM2.5 values,
* calculating rolling average features,
* preserving the chronological order of observations for each monitoring station,
* and producing a feature-rich dataset ready for model training.

```python
import pandas as pd
import numpy as np
```

```python
daily_df = pd.read_csv("lahore_air_quality_cleaned.csv")

print(daily_df.shape)
daily_df.head()
```

```python
daily_df["date"] = pd.to_datetime(daily_df["date"])
```

```python
daily_df = daily_df.sort_values(["name", "date"]).reset_index(drop=True)
```

```python
daily_df["pm25_day1"] = daily_df.groupby("name")["pm25"].shift(-1)
daily_df["pm25_day2"] = daily_df.groupby("name")["pm25"].shift(-2)
daily_df["pm25_day3"] = daily_df.groupby("name")["pm25"].shift(-3)
```

```python
daily_df["pm25_lag1"] = daily_df.groupby("name")["pm25"].shift(1)
daily_df["pm25_lag2"] = daily_df.groupby("name")["pm25"].shift(2)
daily_df["pm25_lag3"] = daily_df.groupby("name")["pm25"].shift(3)
```

```python
daily_df["pm25_7day_avg"] = (
    daily_df.groupby("name")["pm25"]
    .transform(lambda x: x.rolling(7).mean())
)
```

```python
daily_df = daily_df.dropna().reset_index(drop=True)
```

```python
print(daily_df.shape)
daily_df.head()
```

```python
stations = sorted(daily_df["name"].unique())
print(stations)
```

```python
daily_df.to_csv("lahore_air_quality_features.csv", index=False)
```
